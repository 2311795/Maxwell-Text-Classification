```python
# Step 4: Dropdown to select and run any script
import os
import ipywidgets as widgets
from IPython.display import display, clear_output
from datetime import datetime

# Map filenames to readable labels
script_map = {
    "Matrix Response/matrix_decision_making.py": "Matrix: Decision Making",
    
}

dropdown = widgets.Dropdown(
    options=[(label, path) for path, label in script_map.items()],
    description='Run Script:',
    layout=widgets.Layout(width='70%')
)

run_button = widgets.Button(description="▶️ Run", button_style="success")
output = widgets.Output()

def run_script(b):
    with output:
        clear_output()
        selected_script = dropdown.value
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        log_filename = f'output_{timestamp}.txt'
        
        print(f"📄 Running: {selected_script}\nSaving output to: {log_filename}\n")
        os.system(f'python "{selected_script}" | tee "{log_filename}"')

        print("\n✅ Script finished. You can download the log file if needed.")

run_button.on_click(run_script)
display(dropdown, run_button, output)
```
